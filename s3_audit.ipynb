```python
import boto3
```

```python
s3 = boto3.client("s3")
response = s3.list_buckets()
```

```python
for buckets in response["Buckets"]:
    result_acl = s3.get_bucket_acl(Bucket=buckets["Name"])
    print(result_acl)
```

```python
for buckets in response["Buckets"]:
    result_pab = s3.get_public_access_block(Bucket=buckets["Name"])
    print(result_pab)
```
