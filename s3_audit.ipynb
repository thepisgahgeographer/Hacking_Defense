```python
import boto3
```

```python
s3 = boto3.client("s3")
response = s3.list_buckets()
```

```python
for buckets in response["Buckets"]:
    result_acl = s3.get_bucket_acl(Bucket=buckets["Name"])
    print(result_acl)
```

```python
for buckets in response["Buckets"]:
    result_pab = s3.get_public_access_block(Bucket=buckets["Name"])
    print(result_pab)
```

```python
import json

# Create a bucket policy
bucket_name = 'amzn-s3-demo-bucket'
bucket_policy = {
    'Version': '2012-10-17',
    'Statement': [{
        'Sid': 'AddPerm',
        'Effect': 'Allow',
        'Principal': '*',
        'Action': ['s3:GetObject'],
        'Resource': f'arn:aws:s3:::{bucket_name}/*'
    }]
}

# Convert the policy from JSON dict to string
bucket_policy = json.dumps(bucket_policy)

# Set the new policy
s3 = boto3.client('s3')
s3.put_bucket_policy(Bucket=bucket_name, Policy=bucket_policy)
```
